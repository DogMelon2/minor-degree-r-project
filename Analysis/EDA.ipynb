```python
refinery_df["Weight Difference"] = (
    refinery_df["Origin Net Weight"]
    - refinery_df["Destination Net Weight"]
)

refinery_df["Weight Loss %"] = (
    refinery_df["Weight Difference"]
    / refinery_df["Origin Net Weight"]
) * 100

print(refinery_df.head())
```

```python
print(refinery_df[[
    "Origin Net Weight",
    "Destination Net Weight",
    "Weight Difference",
    "Weight Loss %"
]].describe())
```

```python
print("\nAverage weight loss by refinery:")
print(
    refinery_df.groupby("Refinery")["Weight Loss %"]
    .agg(["count", "mean", "min", "max"])
    .sort_values("mean")
)
```

```python
print("\nAverage weight loss by product type:")
print(
    refinery_df.groupby("Type")["Weight Loss %"]
    .agg(["count", "mean", "min", "max"])
    .sort_values("mean")
)
```

```python
print("\nAverage weight loss by work shift:")
print(
    refinery_df.groupby("Work shift")["Weight Loss %"]
    .agg(["count", "mean", "min", "max"])
    .sort_values("mean")
)
```

```python
import jdatetime
import pandas as pd

def jalali_to_gregorian(date_str):
    y, m, d = map(int, date_str.split("/"))
    return jdatetime.date(y, m, d).togregorian()

refinery_df["Departure Date"] = refinery_df["Origin Departure Date"].apply(
    jalali_to_gregorian
)

refinery_df["Arrival Date"] = refinery_df["Destination Arrival Date"].apply(
    jalali_to_gregorian
)

print(refinery_df[
    ["Origin Departure Date", "Departure Date",
     "Destination Arrival Date", "Arrival Date"]
].head())
```

```python
refinery_df["Transit Time (days)"] = (
    refinery_df["Arrival Date"] -
    refinery_df["Departure Date"]
).apply(lambda x: x.days)

print(refinery_df["Transit Time (days)"].describe())
```

```python
print(refinery_df["Transit Time (days)"].value_counts().sort_index())
```

```python
correlation = refinery_df[
    ["Transit Time (days)", "Weight Loss %"]
].corr()

print(correlation)
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=refinery_df,
    x="Transit Time (days)",
    y="Weight Loss %"
)

plt.axhline(0, linestyle="--")

plt.title("Relationship Between Transit Time and Weight Difference")
plt.xlabel("Transit Time (Days)")
plt.ylabel("Weight Difference (%)")

plt.tight_layout()
plt.show()
```
